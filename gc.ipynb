```python

# Python automatically manages memory.
# This notebook demonstrates:
# 1. Reference counting
# 2. Automatic object destruction
# 3. Circular references
# 4. Garbage collector (gc module)
# 5. Generational garbage collection

import gc
```

```python
# This demonstrates reference counting.
# When the reference count reaches zero,
# the object is destroyed immediately.

class Demo:
    def __init__(self, name):
        self.name = name
        print(f"{self.name} created")

    def __del__(self):
        print(f"{self.name} destroyed")

a = Demo("Object A")   # Object created
print("Created at:", hex(id(a)))
print("Deleting reference 'a'")
del a                 # Reference count becomes zero

print("End of cell")
```

```python
# Circular references occur when two or more objects reference each other, creating a cycle.
class Node:
    def __init__(self, name):
        self.name = name
        self.ref = None
        print(f"{self.name} created")

    def __del__(self):
        print(f"{self.name} destroyed")

a = Node("A")
b = Node("B")

a.ref = b
b.ref = a

del a
del b

print("End of program")

## Here is where the garbage collector comes in. It detects circular references and cleans them up.
# If there is no gc , the __del__ method will never be called and the memory will not be freed.
```

```python

# Forcing garbage collection to see the effect of circular references.
import gc

class Node:
    def __init__(self, name):
        self.name = name
        self.ref = None

    def __del__(self):
        print(f"{self.name} at {hex(id(self))} destroyed")

a = Node("A")
b = Node("B")

print("A address:", hex(id(a)))
print("B address:", hex(id(b)))

a.ref = b
b.ref = a

del a, b

print("Forcing GC")
gc.collect()
```

```python
import gc

print("Initial GC counts:", gc.get_count())

a = []
for i in range(10000):
    a.append({i: i*i})
    
print("After creating objects:", gc.get_count())

del a

print("After object deletion:", gc.get_count())

gc.collect()

print("After manual GC:", gc.get_count())
```

```python
# With python 3,14, The cycle garbage collector is now incremental. This means that instead of stopping the program to perform a full garbage collection,
# it can now perform small steps of garbage collection while the program is running, which can help reduce latency and improve performance.

# Example of incremental garbage collection:
import gc
print("Initial GC counts:", gc.get_count())
a = []
for i in range(10000):
    a.append({i: i*i})
print("After creating objects:", gc.get_count())
```
